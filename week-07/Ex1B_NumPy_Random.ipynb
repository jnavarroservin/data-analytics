## Generating Random Numbers

**Create a variable named r_1 that contains a random set of integers between 1 and
20 with a size of 2 elements. Print r_1.**

```python
r_1 = np.random.randint(1, 21, size=2)
print(r_1)
```

This generates two random integers(whole numbers) between 1 and 20. This is also useful in the real world to help generate fake customer_id or fake testing_data. Each time this code is ran the output will re-gerate a different output.

**Create a variable named r_2 that contains a random set of integers between 1 and

```python
r_2 = np.random.randint(1, 21, size=(1, 2))
print(r_2)
```

## Generating Random Choices from an Array

**Create a variable named students and assign it a 1-D array containing the names
of 12 of your classmates**

```python
students = np.array([
    'Cagla', 'Mina', 'Sultan', 
    'Maritza', 'Paola', 'Julissa', 
    'Rodas', 'Godknows', 'Longley', 
    'Onur', 'Ceyhun', 'Jimmy'])
print(students)
```

**Create a variable named call_on and use it to choose one random selection from
students. Print call_on.**

```python
call_on = np.random.choice(students)
print(call_on)
```

**Create a variable named breakout_groups and use it to choose a random

```python
breakout_groups = np.random.choice(students, size=(3, 4))
print(breakout_groups)
```

## Generating a Normal Distribution

**Create a variable named bell_curve and assign it a random distribution for
which the mean (peak of the curve) is 10, the standard deviation is 5, and the size of
the returned array is 4 rows by 5 columns. Print bell_curve**

```python
bell_curve=np.random.normal(10,5,(4,5))
print(bell_curve)
```

## Looping Through Arrays

**Create a variable named arrA as a two-dimensional array containing (11, 12, 13) and (14, 15, 16) then print arrA**

```python
arrA = np.array([
    [11, 12, 13],
    [14, 15, 16]
])

print(arrA)
print(type(arrA))
# nested lists create rows and columns, making it 2D.
# simply having [] creates a singular row list, making it 1D.
# However, [[]] creates a 2D array with one row + one column.
# Python's version of a spreadsheet table.
```

**Loop through the first dimension of arrA (i.e. for x in arrA) and
print each element multiplied by 2.**

```python
# Loop through the first dimension of arrA (i.e. for x in arrA) and
# print each element multiplied by 2.

for x in arrA:
    print(x * 2)

# Each row is x, so I multiply each row by 2, which multiplies each element in the row by 2.
# Also, NumPy already applies the multiplication to each element in the array,
# so I don't have to loop through each row.
```

**Loop through the second dimension of arrA (i.e. for y in x) and
print each element multiplied by 2**

```python
for x in arrA:
    for y in x:
        print(y * 2)

# This loops through each value. Also, the x is the row, and y is the value in the row.
# So, I multiply each value by 2.
```

First loop: works the code row by row, treating the values inside each list as a single unit.
So, each arrA(aka rows) are printed * 2.

Second Row: the code goes one level deeper and loops through each individual value inside the rows.
It prints each number * by 2 separately.

The key difference is that the first loop works at the ROW level (first dimension), while the second loop works at the VALUE level (second dimension), giving more detailed control over each value.

**Again loop through the second dimension of arrA (i.e. for y

## Joining and Stacking Arrays

**Create variable joined_array_1 and assign it the value of arrA concatenated
with arrA on axis=0 (the default axis). Print joined_array_1**

```python
joined_array_1 = np.concatenate((arrA, arrA), axis=0)
print(joined_array_1)
```

The axis is by default 0, so it doesnt necessarily need to be added but it helps me practice.
Also, concatenating arrays helps "glue" multiple arrays together.

(axis = 0) helps the code JOIN the arrays vertically by rows.
(axis = 1) helps the code JOIN the arrays horizontally by columns.

**Create variable joined_array_2 and assign it the value of arrA concatenated
with arrA on axis=1. Print joined_array_2.**

```python
joined_array_2 = np.concatenate((arrA, arrA), axis=1)
print(joined_array_2)
```

Now, (axis=1) joins the arrays horizontally, adding more columns to each row elongating is sideways.

**Create variable stacked_array_1 and assign it the value of joined_array_1
stacked with joined_array_1 on the default axis. Print stacked_array_1.**

```python
stacked_array_1 = np.stack((joined_array_1, joined_array_1), axis=0)
print(stacked_array_1)
```

Stacking combines arrays along a new axis, adding an extra dimension (layers) to the output.

## Splitting Arrays

**Create variable splitsky and assign it the value of joined_array_1 split into 2.
Print the variable. Then add a comment to the cell to describe the result: how many
arrays? Of how many dimensions each?**

```python
splitsky = np.array_split(joined_array_1, 2)

print(splitsky)
```

The `array_split()` divides `joined_array_1` into two separate arrays each resulting array is still 2D because they contain rows AND columns. (Think of it as an excel sheet)

 ## Searching and Replacing Array Values

**Create variable result_1. For the value of this variable, use where() to find each even number within arrA (hint, you’ll need to write a condition using %
and ==). If the condition is met, print the value from arrA; if not met, print 'odd'.**

```python
result_1 = np.where(arrA % 2 == 0, arrA, 'odd')
print(result_1)
```

If a value in arrA is even, it keeps the original number, but if the value is odd, it gets replaced with the word 'odd'.
Basically reads as: `np.where(condition, value_if_true, value_if_false)`

## Sorting and Filtering Arrays

**Create variable arrB as a 2-D array including the values (7,6), (13,5), and (7,3).
Create another variable, arrB_sorted, and assign it the sorted value of arrB.
Print arrB_sorted.**

```python
arrB = np.array([
    (7, 6),
    (13, 5),
    (7, 3)])
arrB_sorted = np.sort(arrB)
print(arrB_sorted)
```

The `np.sort()` sorts the values in each row from smallest to largest, but it only sorts by row not the entire list.

**Finally, let’s filter arrB using conditional logic. Start the cell with a print statement
to again print the contents of arrB. Then create variable z with the value arrB < 7
to look for values less than 7. Next create variable arrC, and assign it arrB[z].
Print arrC**

```python
print(arrB)

z = arrB < 7

arrC = arrB[z]

print(arrC)
```

The z stores True/False values depending if each number is < 7. The `arrB[z]` filters the array and keeps only the values that match the condition I ask for is True.

**Applying the same sort of structure, how would you create a new conditional logic
filter y to look for values greater than or equal to 7? Follow similar steps as above to
create a new variable arrD to hold the filtered results, and print arrD**

```python
y = arrB >= 7

arrD = arrB[y]

print(arrD)
```

The  y stores True/False values for numbers >= 7. The `arrD` keeps only the values that match the condition, so it only keeps `True` values.

How its read: `condition = array comparison
filtered_array = array[condition]`

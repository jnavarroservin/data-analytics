4. WORKPLACE EXAMPLE: running experiments without modifying the original data

**Show what happens if you add np.array([1]) to r_4.**

```python
np.add(np.array([1]), r_4)
```

This works because the array [1] contains a single value. NumPy broadcasts that one value across every element in r_4.

Since a single value can stretch to match any array shape, 1 gets added to every number in r_4.

**Show what happens if you add np.array([[1]]) to r_4.**

```python
np.add(np.array([[1]]), r_4)
```

This also works because [[1]] is a 2-dimensional array with shape (1,1). NumPy can still broadcast this single value across all rows and columns of r_4.

Even though it is technically 2D, it still only contains one value so NumPy stretches it across the entire array.

**Show what happens if you add np.array([1,1,1,1,1]) to r_4.**

```python
np.add(np.array([1,1,1,1,1]), r_4)
```

This works because the array has 5 values, which matches the 5 columns in r_4.

NumPy broadcasts the row across each row of r_4 and adds the values column-by-column. Since the dimensions align correctly, no error occurs.

**What happens if you add np.array([[1],[1]]) to r_4? Add a comment to the
cell with the error message this produces.**

```python
np.add(np.array([[1],[1]]), r_4)
```

This returns an error because the shapes (2,1) and (4,5) do not match for broadcasting.

The 1 column can stretch to fit 5 columns, but 2 rows cannot stretch to fit 4 rows, so NumPy cannot add the arrays together.

**Explain how broadcasting works to combine the values of two arrays. Using the add function, what gets added to what?
What determines the shape of the resulting array?**

Broadcasting lets NumPy combine arrays of different shapes by stretching the smaller array to fit the bigger one when possible.

The add() function allows the values to be added together element by element. If one array is smaller, NumPy repeats or stretches those values across the larger array.

The final shape usually matches the larger array, as long as the dimensions are compatible.

## Grocery Problem

**Create two new variables, bag1 and bag2. The first variable represents the
contents of a grocery bag, which contains two vegetables, three fruits, four boxes of
snacks, and no beverages. The second variable represents another grocery bag,
which contains no fruits, four vegetables, two beverages, and one box of snacks.
Assign an appropriate array to each variable that represents how many of each
category of item are in each bag. Print bag1 and bag2**

```python
bag1 = np.array([2, 3, 4, 0])
bag2 = np.array([4, 0, 1, 2])

print(bag1)
print(bag2)
# match the total number of items in each bag to the correct category index.
```

```python
print(f'The first bag contains {np.sum(bag1)} items, and the second bag contains {np.sum(bag2)} items.')
```

**Write an expression to calculate the total number of items in each bag, and print the
result in the form of a sentence: The first bag contains __ items, and the second bag
contains ___ items.**

```python
totals = np.vstack((bag1,bag2))
print(totals)
```

**Write an expression to calculate the total number of items in each category, and
print the result in the form of a sentence: There are __ vegetables, __ fruits, __ boxes
of snacks, and __ beverages in the two grocery bags.**

```python
category_totals = np.sum(totals, axis=0)
print(f'There are {category_totals[0]} vegetables, {category_totals[1]} fruits, {category_totals[2]} boxes of snacks, and {category_totals[3]} beverages in the two grocery bags.')
```

## Teamwork

**There are three people on your team: Ali, Ben, and Carla. Your squad went out to knock on doors each evening
last week, Monday through Friday, and now you are assembling that data.**

**Ali reported knocking on 5 doors on Monday, 4 on Tuesday, 6 on Wednesday, 4 on
Thursday, and 2 on Friday. Create a variable holding an array of these values.**

```python
Ali = np.array([5, 4, 6, 4, 2])
```

- doors create a singular row
- vertically there's 5 columns for each day a door was knocked on

**Ben missed Monday because he was out of town. He reported knocking on 5 on
Tuesday, 5 on Wednesday, 3 on Thursday, and 4 on Friday. Create a variable holding
an array of Ben’s value**

```python
Ben = np.array([0, 5, 5, 3, 4])
```

**Carla reported knocking on 3 doors on Monday, 6 on Tuesday, 3 on Thursday, and 4 on Friday. She was sick on Wednesday so missed that day. Create a variable holding an array of Carla’s values**

```python
Carla = np.array([3, 6, 0, 3, 4])
```

**Compile the values reported by Ali, Ben, and Carla into a single array containing all door-knocks for the week, as all_knocks. Print the resulting array**

```python
all_knocks = (np.array([Ali, Ben, Carla]))
# all_knocks = Ali + Ben + Carla
print(all_knocks)
```

**Use all_knocks to calculate the total number of doors that each person knocked
on for the week, and also the total number of doors that the group collectively
knocked on. Print the results**

```python
print(f'Ali knocked on {np.sum(Ali)} doors, Ben knocked on {np.sum(Ben)}, Carla knocked on {np.sum(Carla)}.')
print(f'Total doors knocked: {np.sum(all_knocks)}')
```

**The organization you’re working with estimates that for every two door-knocks, one
of those people will actually turn out to vote. For your report to the organization,
what is the rolling total of estimated votes that your group was able to turn out, day
by day for last week? (Hint: Start with the rolling total by day, then multiply that array
by an appropriate value to get the result.**

```python
daily_total = np.sum(all_knocks, axis=0)
rolling = np.cumsum(daily_total)
estimated_votes = rolling * 0.5
print(estimated_votes)
```

```python
# What other number or string value(s) can be typecast as boolean values? Use the print() function to list them
# INTEGERS:
print(bool(1))
# Output: true - Python reads 1 as True
print(bool(0))
# Output: false - Python reads 0 as False

# FLOATS:
print(bool(5.12))
# Output: true - the output printed is an integer
print(bool(0.0))
# Output: false - the output printed has a value of zero

# STRINGS:
print(bool('red'))
# Output: true - the output printed is a true string with characters inside
print(bool(''))
# Output: false - the outout printed has no character so it's false and doesnt really print anything

# In Python:
# Anything empty, zero, or nothing = False
# Everything else = True
```

```python
# What's the difference between 0 and 0.0?  Write your answer below as a code comment.
print(type(0))      # <class 'int'>
print(type(0.0))    # <class 'float'>

# 0 is an integer and 0.0 is a float.
# They represent the same value mathematically,
# but Python treats them as different data types.
# 0 is a whole number with no decimal point.
# 0.0 has a decimal point, meaning it can represent partial values.
# Think of it like the difference between saying
# "I have 0 dollars" vs "I have 0.00 dollars" —
# same amount, but one is more precise in how it's expressed.
```
